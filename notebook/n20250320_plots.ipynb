```python
import numpy as np
import matplotlib.pyplot as plt

# Sample bid and ask data with some "holes" (missing price levels)
bid_prices = np.array([100.5, 100.4, 100.3, 100.1])  # Missing 100.2
ask_prices = np.array([100.6, 100.7, 100.8, 101.0])  # Missing 100.9

# Corresponding volumes for bids and asks
bid_volumes = np.array([10, 20, 30, 50])
ask_volumes = np.array([15, 25, 35, 55])

# Create a full price range from min to max of bid and ask prices
all_prices = np.linspace(min(bid_prices.min(), ask_prices.min()), max(bid_prices.max(), ask_prices.max()), num=50)

# Create volume arrays for each price level (initialize with 0)
bid_hist = np.zeros_like(all_prices)
ask_hist = np.zeros_like(all_prices)

# Fill in volumes at the bid and ask price levels
for i, price in enumerate(bid_prices):
    bid_hist[np.abs(all_prices - price).argmin()] = bid_volumes[i]

for i, price in enumerate(ask_prices):
    ask_hist[np.abs(all_prices - price).argmin()] = ask_volumes[i]

# Plotting the histogram
plt.figure(figsize=(10, 6))

# Plotting the bid volumes (green bars)
plt.bar(all_prices, bid_hist, width=(all_prices[1] - all_prices[0]), color='green', alpha=0.6, label='Bids')

# Plotting the ask volumes (red bars)
plt.bar(all_prices, ask_hist, width=(all_prices[1] - all_prices[0]), color='red', alpha=0.6, label='Asks')

# Labels and title
plt.xlabel('Price')
plt.ylabel('Volume')
plt.title('Order Book Snapshot')

# Adding the legend
plt.legend()

# Show grid for better visualization
plt.grid(True)

# Show the plot
plt.show()
```
